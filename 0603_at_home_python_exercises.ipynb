## 1 · Basic Python

### Recap

| Concept | Key idea | Quick example |
|---|---|---|
| **Variables & data types** | A variable stores a value. Common types: `int`, `float`, `str`, `bool`. | `x = 42` |

## 2 · NumPy

### Recap

NumPy is the core library for numerical computing in Python. Its main object is
the **ndarray** — a fast, fixed-type, n-dimensional array.

```python
import numpy as np

a = np.array([1, 2, 3])          # 1-D array from a list
b = np.zeros((2, 3))             # 2×3 array of zeros
c = np.arange(0, 10, 2)          # [0, 2, 4, 6, 8]
```

| Operation | Example |
|---|---|
| **Indexing / slicing** | `a[0]`, `a[1:3]`, `b[0, :]` |
| **Element-wise arithmetic** | `a + 10`, `a * b`, `a ** 2` |
| **Aggregations** | `a.mean()`, `a.sum()`, `np.std(a)` |
| **Shape & dimensions** | `a.shape`, `a.ndim`, `a.reshape(3, 1)` |

## 3 · Matplotlib

### Recap

`matplotlib.pyplot` (commonly imported as `plt`) is the standard plotting
library in Python.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]
plt.plot(x, y, label="sales")
plt.scatter(x, y, color="red")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Quarterly Sales")
plt.legend()
plt.grid(True)
plt.show()
```

| Function | Purpose |
|---|---|
| `plt.plot()` | Line plot |
| `plt.scatter()` | Scatter plot |
| `plt.xlabel()` / `plt.ylabel()` | Axis labels |
| `plt.title()` | Chart title |
| `plt.legend()` | Show legend (needs `label=` in plot calls) |
| `plt.grid(True)` | Toggle grid lines |

Pandas provides two core structures: **Series** (1-D) and **DataFrame** (2-D
table). A DataFrame is essentially a dictionary of Series sharing the same
index.


| **Detect missing values** | `df.isna()`, `df.isna().sum()` |
