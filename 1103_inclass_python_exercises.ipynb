### Recap

Pandas provides two core structures: **Series** (1-D) and **DataFrame** (2-D
table). A DataFrame is essentially a dictionary of Series sharing the same index.

```python
import pandas as pd

df = pd.DataFrame({
    "name":  ["Alice", "Bob", "Carol"],
    "score": [85, 92, 78]
})
```

| Task | Code |
|---|---|
| **Filter rows** | `df[df["col"] > value]` |
| **Add a column** | `df["new_col"] = values` |
| **Group & aggregate** | `df.groupby("col")["val"].mean()` |
| **Sort** | `df.sort_values("col", ascending=False)` |
| **Remove duplicates** | `df.drop_duplicates()` |
| **Detect missing** | `df.isna().sum()` |
| **Fill missing** | `df["col"].fillna(df["col"].median())` |
| **Apply function** | `df["col"].apply(lambda x: ...)` |
| **Pivot table** | `df.pivot_table(values=..., index=..., columns=...)` |
