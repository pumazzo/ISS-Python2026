## 0. Lists and mutability reminder

A Python list is a mutable sequence. *In-place* operations modify the existing list object
instead of creating a new one.

Key idea:
- In-place operations change the list without changing its identity (`id(list_obj)`).
- Non in-place operations return a new list (different `id`).

````python
numbers = [1, 2, 3]
print("Before:", numbers, "id:", id(numbers))

# In-place: append
numbers.append(4)
print("After append:", numbers, "id:", id(numbers))  # same id
````

## 1. `append` and `extend`

- `list.append(x)`: add a single element `x` at the end (in-place).
- `list.extend(iterable)`: add all elements from `iterable` at the end (in-place).

````python
data = [1, 2]
print("Start:", data)

# append: adds a single element
data.append(3)
print("After append(3):", data)

# extend: adds elements from an iterable
data.extend([4, 5])
print("After extend([4, 5]):", data)

# extend also works with other iterables, e.g. range
data.extend(range(6, 9))
print("After extend(range(6, 9)):", data)
````

## 2. `insert`, `pop`, `remove`, and `clear`

- `list.insert(index, value)`: insert `value` at position `index` (in-place).
- `list.pop([index])`: remove and return element at `index` (default: last) (in-place).
- `list.remove(value)`: remove first occurrence of `value` (in-place).
- `list.clear()`: remove all elements (in-place).

````python
items = [10, 20, 30, 40]
print("Start:", items)

# insert
items.insert(2, 25)  # insert at index 2
print("After insert(2, 25):", items)

# pop last element
last = items.pop()
print("Popped last:", last, "Remaining:", items)

# pop element at index 1
removed = items.pop(1)
print("Popped index 1:", removed, "Remaining:", items)

# remove first occurrence of a value
items.remove(25)
print("After remove(25):", items)

# clear the list
items.clear()
print("After clear():", items)
````

## 3. `sort` and `reverse`

- `list.sort(key=None, reverse=False)`: sort the list in-place and returns `None`.
- `list.reverse()`: reverse the elements in-place.

Important: `sorted(list)` returns a new sorted list, while `list.sort()` modifies the list itself.

````python
values = [5, 3, 1, 4, 2]
print("Original:", values)

# in-place sort
result = values.sort()
print("After sort():", values)
print("Return value of sort():", result)

# reverse in-place
values.reverse()
print("After reverse():", values)

# contrast with sorted()
values = [5, 3, 1, 4, 2]
sorted_values = sorted(values)
print("values after sorted():", values)
print("sorted(values):", sorted_values)
````

## 4. Slice assignment and deletion

Slice assignment is another way to modify a list in-place:

- `lst[i:j] = iterable` replaces the slice with the elements of `iterable`.
- `del lst[i:j]` deletes the slice.
- `del lst[index]` deletes a single element.


````python
lst = [0, 1, 2, 3, 4, 5]
print("Start:", lst)

# replace a slice
lst[2:4] = [20, 30, 40]
print("After lst[2:4] = [20, 30, 40]:", lst)

# delete a slice
del lst[1:3]
print("After del lst[1:3]:", lst)

# delete a single element
del lst[0]
print("After del lst[0]:", lst)
````

## 5. In-place vs. not in-place

Some operations look similar but behave differently:

- `lst += [x, y]` modifies `lst` in-place (for lists).
- `lst = lst + [x, y]` creates a new list and rebinds `lst`.
- `sorted(lst)` creates a new list, `lst.sort()` modifies in-place.

You can check this by printing `id(lst)` before and after the operation.

````python
lst = [1, 2, 3]
print("Original lst:", lst, "id:", id(lst))

# Using += (in-place for lists)
lst += [4, 5]
print("After lst += [4, 5]:", lst, "id:", id(lst))

# Using + (new list)
lst = lst + [6, 7]
print("After lst = lst + [6, 7]:", lst, "id:", id(lst))
````

## 6. Modifying lists while iterating

Modifying a list in-place while iterating over it can lead to unexpected behavior.
A common pattern is to iterate over a copy or build a new list instead.


````python
nums = [1, 2, 3, 4, 5, 6]
print("Original:", nums)

# Example of a problematic pattern: removing while iterating
for x in nums:
    if x % 2 == 0:
        nums.remove(x)

print("After removing evens in-loop:", nums)

# Safer pattern: build a new list
nums = [1, 2, 3, 4, 5, 6]
filtered = [x for x in nums if x % 2 != 0]
print("Filtered without modifying in-loop:", filtered)
print("Original unchanged:", nums)
````
