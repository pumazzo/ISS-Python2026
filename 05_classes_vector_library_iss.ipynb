## Functions – Minimal Recap

Functions group reusable logic into named, reusable blocks.

```python
def add(a, b):
    return a + b


def square(x):
    return x * x
```

### Parameters and Return Values

* Parameters are inputs listed inside parentheses.
* `return` sends a result back to the caller.
* If no `return` is specified, the function returns `None`.

Example usage:

```python
x = add(3, 4)
y = square(x)
```

### Type Hints

Type hints specify the expected types of parameters and return values.

Syntax:

* `name: Type` for parameters or variables
* `-> Type` for return values

Examples:

```python
def greet(name: str) -> str:
    return "Hello " + name


def average(values: list[float]) -> float:
    return sum(values) / len(values)
```

```python
def add(a: int, b: int) -> int:
    return a + b


def square(x: int) -> int:
    return x * x
```

Type hints:

* Improve readability.
* Help static analysis tools (e.g., type checkers).
* Are **not enforced at runtime**.

### Good Practice

* Keep functions short and focused on a single task.
* Use clear parameter names.
* Use type hints to document expected inputs and outputs.
* Avoid duplicating logic: extract repeated code into functions.


````python
# Function with default parameter
def power(base: float, exponent: int = 2) -> float:
    return base ** exponent

print(power(3))        # 9.0
print(power(3, 3))     # 8.0


# Function returning multiple values
def min_max(values: list[int]) -> tuple[int, int]:
    return min(values), max(values)

lo, hi = min_max([3, 1, 7, 4])
print(lo, hi)          # 1 7



# function inside a function
def square(x: int) -> int:
    return power(x,2)

# Function as argument (higher-order function)
def apply_twice(f, x: int) -> int:
    return f(f(x))

print(apply_twice(square, 3))   # 81
````

Special methods begin and end with double underscores.
They define how objects behave with built-in operations.

* `__repr__` → string representation
* `__len__` → behavior of `len(obj)`

## Vector Algebra, Implemented Features

This notebook implements a minimal subset of vector algebra in $( \mathbb{R}^n )$ using a custom `Vector` class written in pure Python.

A vector is defined as an ordered list of real components:

$
\mathbf{v} = (v_1, v_2, \dots, v_n)
$

### 1. Vector Addition

Two vectors of the same dimension can be added component-wise:

$
\mathbf{u} + \mathbf{v} = (u_1 + v_1, \dots, u_n + v_n)
$

This operation is implemented via the `+` operator. A `TypeError` is raised if the dimensions differ.

### 2. Scalar Multiplication

A vector can be multiplied by a real scalar:

$
\alpha \mathbf{v} = (\alpha v_1, \dots, \alpha v_n)
$

This is implemented using the `*` operator (only `v * scalar`).

### 3. Scalar (Dot) Product

The scalar product between two vectors of the same dimension is defined as:

$
\mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^{n} u_i v_i
$

It is available both as:

* the method `v.dot(w)`
* the operator `v @ w`

The result is a floating-point number.

### 4. Euclidean Norm

The Euclidean norm (modulus) of a vector is:

$
|\mathbf{v}| = \sqrt{\sum_{i=1}^{n} v_i^2}
$

It is exposed as a read-only property `norm` and computed on demand.

### Design Notes

* Components are stored internally as a list of `float`.
* Individual components can be modified by index.
* No external libraries (e.g., NumPy) are used.

The implementation is intentionally minimal to illustrate core object-oriented programming concepts such as encapsulation, operator overloading, and properties.


# A tiny vector library (pure Python)

This notebook is a compact example for practicing **classes and objects**.

We implement a `Vector` object that supports:
- Vector addition with `+` (only for vectors of the same dimension)
- Scalar multiplication with `*` (only `v * scalar` for simplicity)
- Scalar (dot) product as both:
  - a method: `v.dot(w)`
  - an operator: `v @ w`

Design constraints (for teaching purposes):
- Components are stored internally as a **Python list** of `float`.
- Components can be updated by index (e.g., `v[0] = 3.5`) but the **dimension is fixed** once created.
- `dimension` is read-only and cannot be changed after construction.
- `norm` is a property computed on the fly (no caching).


````python
from __future__ import annotations


from math import sqrt
from typing import Iterable, Iterator, List, Union

Number = Union[int, float]
````

````python
class Vector:
    """A simple mathematical vector in R^n (pure Python).

    - Stores components as a list of float.
    - Dimension is fixed at construction time.
    - Components can be updated by index, but resizing is intentionally not supported.
    """

    def __init__(self, components: Iterable[Number]):
        comps = list(components)
        if len(comps) == 0:
            raise ValueError("Vector must have at least one component.")
        # Cast to float at instance time
        self._components: List[float] = [float(x) for x in comps]
        self._dimension: int = len(self._components)

    @property
    def dimension(self) -> int:
        """Fixed dimension (read-only)."""
        return self._dimension

    @property
    def norm(self) -> float:
        """Euclidean norm (modulus), computed on the fly."""
        return sqrt(sum(x * x for x in self._components))

    @property
    def components(self) -> List[float]:
        """A copy of the internal components (softly prevents resizing)."""
        return list(self._components)

    def __len__(self) -> int:
        return self._dimension

    def __iter__(self) -> Iterator[float]:
        return iter(self._components)

    def __getitem__(self, idx: int) -> float:
        return self._components[idx]

    def __setitem__(self, idx: int, value: Number) -> None:
        # Allow component updates, but by index only (no resizing API).
        self._components[idx] = float(value)

    def __repr__(self) -> str:
        return f"Vector({self._components})"

    def _require_same_dimension(self, other: object) -> "Vector":
        if not isinstance(other, Vector):
            raise TypeError(f"Expected a Vector, got {type(other).__name__}.")
        if self.dimension != other.dimension:
            raise TypeError(
                f"Dimension mismatch: {self.dimension} vs {other.dimension}."
            )
        return other

    def __add__(self, other: object) -> "Vector":
        other_v = self._require_same_dimension(other)
        return Vector(a + b for a, b in zip(self._components, other_v._components))

    def __mul__(self, scalar: object) -> "Vector":
        # Simple example: only v * scalar (not scalar * v)
        if not isinstance(scalar, (int, float)):
            raise TypeError(
                f"Can only multiply Vector by a scalar (int/float), got {type(scalar).__name__}."
            )
        return Vector(x * float(scalar) for x in self._components)

    def dot(self, other: object) -> float:
        other_v = self._require_same_dimension(other)
        return sum(a * b for a, b in zip(self._components, other_v._components))

    def __matmul__(self, other: object) -> float:
        # v @ w as dot product
        return self.dot(other)
````

## Demos / quick tests

The following cells show typical usage and a few `assert` statements that act like lightweight tests.


````python
# Create vectors
v = Vector([1, 2, 3])
w = Vector([10, 20, 30])

print("v =", v)
print("w =", w)
print("dimension(v) =", v.dimension)
print("norm(v) =", v.norm)
````

````python
# Addition
u = v + w
print("v + w =", u)

# Dot product (two ways)
print("v.dot(w) =", v.dot(w))
print("v @ w     =", v @ w)
````

````python
# Scalar multiplication (v * scalar only)
s = v * 2.5
print("v * 2.5 =", s)
````

````python
# Component update by index (allowed)
v[0] = 99
print("after v[0] = 99:", v)
print("dimension still fixed:", v.dimension)
print("new norm:", v.norm)
````

````python
# Error handling demos
try:
    Vector([])
except Exception as e:
    print("Empty vector ->", repr(e))

try:
    Vector([1, 2]) + Vector([1, 2, 3])
except Exception as e:
    print("Dimension mismatch ->", repr(e))

try:
    Vector([1, 2, 3]) * "oops"
except Exception as e:
    print("Bad scalar ->", repr(e))
````
