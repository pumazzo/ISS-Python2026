# Control flow constructs

## What is control flow
When programming, think of your code like a road trip that has to be followed. Control flow is simply the __map__ that tells Python where to go next.

There are only 3 ways the "traffic" can move and be influenced in your code:

__Sequence__: Go straight ahead (first do step A, then do step B).

__Selection__: A fork in the road (if X is true, go Left; otherwise, go Right).

__Cycle (Loop)__: Driving around a roundabout (keep doing this until I say stop).

These control structures forms the basis of a programming paradigm called *structured programming*.

## Sequence

## Selection

Sometimes your code needs to make a decision. This is called Selection (or Branching).

Imagine you are at a fork in the road. You ask a question: "Is it raining?"

YES (True): Take the umbrella.

NO (False): Leave the umbrella.

In Python, we use the if-else statement to create this fork.

```python
raining = True
#raining = False
print("Do I have to take the umbrella?")
if raining:
  print("Yes")
else:
  print("No")
```

Or if we want a numeric example:

```python
a, b = 1, 2 #let's try to print the biggest number
```

```python
if a > b:
    print(a)
else:
    print(b)
```

As you can see from the example above, the instructions which are part of the same branch are aligned on the right of the __if__ statement and are considered *inside* the "if branch", same with __else__.

## Indentation

This is the most important rule in Python, something that differs from other programming languages.

How does Python know which instructions belong to one path and which belong to another path? It looks at the empty space at the start of the line, looking at the __indentation__

To indent, we press the space key 4 times, creating some space on the left. This is very important!

This means that __if a line is indented (pushed to the right), it is inside the block above it__.

## Indentation

How much you should indent?

In a way that is consistent within the block! The common sense rules are:
* indent always with 4 spaces
* NEVER use tabs to indent

## Loops
Loops are a way I have to iterate on python objects. We will focus mainly on the __for__ loop.

This loop allows you to iterate over every "iterable" object in python, such as lists, tuples, sets, dicts, range... every objects that contains something.

## for loop

```python
for i in range(3): #in this case, I know how many iterations I have to perform. So i define an object range of 3
                  #(which produces 3 numbers starting from 0 up to 2, cause 0, 1, 2 is 3 numbers and we start counting from 0)
    print(i)      #and I print them
```

## for loop

```python
t = ('a', 'b', 10, 5.5) #in this case I am defining a tuple of 4 elements (a tuple is a fixed list we can't change anymore)
for i in t: #and im telling python: for every element in t, print the element
    print(i)
```

## for loop

```python
d = {'a': 0, 'b': 1, 'c': 2}
for key in d.keys(): #here we use d.keys() (another iterable) to iterate on the dict's keys
                      #we can also, only for keys, write "for key in d" omitting the #.keys()
    print(key)
```

## for loop

```python
for val in d.values(): #same but with the values associated
    print(val)
```

## for loop

```python
for key, val in d.items(): #same but here we are iterating over the pair key-value, so every iteration we can choose
                            #to print one of the to or both of them
    print(key, '=', val)
```

```python
for i in range (10):
  if i % 2 == 0:
    continue  #this way we tell the code to skip that iteration and go to the next one
  print (i)
```

If the body of the `if/for` block contains a single statement, this can be written on the same line:

```python
if a > 10: pass
```

Nothing changes; and is usually better to __avoid that__ as it may trick us.

### But what if I want to iterate as long as some condition is True/False? And I don't know beforehand how long that will persist?

## While loop

The while loop is the generic loop with one condition, while that specific condition is True, the loop will keep iterating.

__REMEMBER__ when the condition STOPS being True, there is a last iteration and then it will exit:

```python
i = 0
while i < 4:
    print(i)
    i += 1
    if i > 2:
      break;  #this way we are telling python "if the condition happens just exit the cycle definitely"
```

## Conversion to bool

The built-in types are automatically converted to bools, for example they can be directly used in conditional expressions. The rules are:

* An `int` equal to 0 is equivalent to False, otherwise `True`
* A `float` equal to 0.0 is equivalent to False, otherwise `True`
* An empty string "" is equivalent to False, otherwise `True`
* An empty container (`[]`, `()`, `set()`, `{}`, ...) is equivalent to False, otherwise `True`
* `None` is equivalent to `False`
